# Case C: all compounds, regimes 2a and 2b

As case B, but with all seven dry compounds (SO4, OC, BC, DU, SS, NO, NH) and a second aerosol population in regime 2b. This is not a realistic aerosol; it makes sure every compound and both regimes go through the code.

The a-bins are mostly soluble (by volume 30 % SO4, 25 % OC, 15 % SS, 10 % NO, 10 % NH, 5 % BC, 5 % DU), in the same two modes as cases A and B. Below 50 nm (regime 1a) only SO4 and OC are used. The b-bins are mostly insoluble (40 % BC, 40 % DU), in one mode of 10 cm⁻³ at 0.5 µm.

Activated b-particles go to their own cloud bins (cloud bins 7–13). As in case B, there is only a trace of rain and no evaporation in 400 s.
