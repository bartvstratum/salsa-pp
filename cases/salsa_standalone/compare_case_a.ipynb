# Case A: old SALSA-standalone test

Rising air parcel. T and p are prescribed every second from `input.dat` (output of a cloud parcel model); RH is 98 % at t = 0. The initial aerosol is sulphate in two modes: 100 cm⁻³ at 0.15 µm and 1 cm⁻³ at 1 µm.

As in the old test, only condensation and the size distribution update are on: no activation, coagulation or autoconversion. Once the air is supersaturated (around t = 200 s), the particles above about 80 nm dry diameter grow into droplets of 10–50 µm. Without activation they stay in the aerosol bins, so the cloud and rain bins remain empty. Condensation only adds water: the dry size, and with it the bin of each particle, does not change.
