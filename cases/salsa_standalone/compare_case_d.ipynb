# Case D: up and down, all paths

Same aerosol and processes as case C, but the parcel goes up and comes back down, so that every phase-1 path is used, including rain and evaporation back to aerosol.

The forcing uses the rows of `input.dat`: up for 2000 s, 1000 s at the top (`input.dat` is constant there), then the same rows backwards to the start, 6000 s in total. RH at the start is 95 % instead of 98 %: at 98 % the rain drops end at 12.6 µm, just above the 10 µm below which SALSA moves rain back to aerosol.

Cloud forms at about 920 s. Rain grows mostly on the way down (from about 3100 s, largest at 4800 s). Between 4950 and 5420 s the cloud bins of both regimes evaporate back to aerosol, and at 5718 s the smallest rain bin does. At that moment both regimes have particles of the right dry size, so the choice between regime a and b (`calc_correlation`) is used as well; it picks a.

Run `python3 salsa_standalone_input.py` first, then `salsa_0d` in `case_d/` (6000 steps, about 3 s, 21 MB output).
